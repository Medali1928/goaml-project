```python
data = [[1, 'Joe', 70000, 3], [2, 'Henry', 80000, 4], [3, 'Sam', 60000, None], [4, 'Max', 90000, None]]
employee = pd.DataFrame(data, columns=['id', 'name', 'salary', 'managerId']).astype({'id':'Int64', 'name':'object', 'salary':'Int64', 'managerId':'Int64'}) 
```

```python
employee
```

```python
pd.merge(
    left=employee,
    right=employee,
    left_on='managerId',
    right_on='id',
    suffixes=('_emp', '_mng')
).loc[lambda row : row['salary_emp'] > row['salary_mng'], ['name_emp']].rename(columns={'name_emp':'Employee'})
```

```python
merged_df = pd.merge(
    left=employee,
    right=employee,
    left_on='managerId',
    right_on='id',
    suffixes=('_emp', '_mng')
)
```

```python
merged_df['salary_emp'] > merged_df['salary_mng']
```

```python
merged_df[merged_df['salary_emp'] > merged_df['salary_mng']]
```

```python
data = [[1, 'a@b.com'], [2, 'c@d.com'], [3, 'a@b.com']]
person = pd.DataFrame(data, columns=['id', 'email']).astype({'id':'Int64', 'email':'object'})
```

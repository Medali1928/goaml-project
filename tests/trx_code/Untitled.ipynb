```python
trx_code = pd.read_excel('accountant_trx.xlsx')
trx_code['TRANSACTION_CODE_ATB'] = trx_code['TRANSACTION_CODE_ATB'].astype(str).str.zfill(3)
```

```python
trx_code_eq_t24 = pd.read_csv('TRANSACTION_CODE_EQA_T24.txt', sep='\t', engine='python')
trx_code_eq_t24['TCDEQU'] = trx_code_eq_t24['TCDEQU'].astype(str).str.zfill(3)
```

```python
pd.merge(
    left=trx_code.query("(ACCOUNTANT_SOURCE == 'EQ') & (OP_TYPE == 'Caisse') & (APP_SOURCE == 'Teller') & (APP_SOURCE_CODE == 'T24TL')"),
    right=trx_code_eq_t24,
    left_on='TRANSACTION_CODE_ATB',
    right_on='TCDEQU',
    indicator=True
).T
```

```python
r = pd.merge(
    left=trx_code.query("(ACCOUNTANT_SOURCE == 'EQ') & (OP_TYPE == 'Caisse') & (APP_SOURCE == 'Teller') & (APP_SOURCE_CODE == 'T24TL')"),
    right=trx_code_eq_t24,
    left_on='TRANSACTION_CODE_ATB',
    right_on='TCDEQU',
    how='inner',
    indicator=True
)
```

```python
r.columns
```

```python
r['ACCOUNTANT_SOURCE'] = 'T24'
r['APP_SOURCE_CODE'] = 'TT'

r.drop(columns=['TRANSACTION_CODE_ATB'], inplace=True)

r.rename(columns={'TCD24': 'TRANSACTION_CODE_ATB'}, inplace=True)
```

```python
r[['ACCOUNTANT_SOURCE', 'APP_SOURCE_CODE', 'TRANSACTION_CODE_ATB',
       'TRANSACTION_DESC_ATB', 'APP_SOURCE', 'OP_TYPE',
       'Categorie_de_transaction', 'TRANSACTION_CODE_GOAML',
       'TRANSACTION_DESC_GOAML', 'FUND_CODE_1', 'FUND_CODE_2']]
```

```python
yy = pd.concat([
    trx_code,
    r[['ACCOUNTANT_SOURCE', 'APP_SOURCE_CODE', 'TRANSACTION_CODE_ATB',
       'TRANSACTION_DESC_ATB', 'APP_SOURCE', 'OP_TYPE',
       'Categorie_de_transaction', 'TRANSACTION_CODE_GOAML',
       'TRANSACTION_DESC_GOAML', 'FUND_CODE_1', 'FUND_CODE_2']]
], ignore_index=True)
```

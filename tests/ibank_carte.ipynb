```python
accountant_card = pd.merge(
    left=transaction_router.eq_accountant.accountant_transactions,
    right=transaction_router.transaction_code[['TRANSACTION_CODE_ATB', 'TRANSACTION_DESC_ATB', 'APP_SOURCE_CODE', 'APP_SOURCE', 'OP_TYPE', 'Categorie_de_transaction', 'TRANSACTION_DESC_GOAML', 'TRANSACTION_CODE_GOAML', 'FUND_CODE_1', 'FUND_CODE_2']],
    left_on=['APP_SOURCE_CODE', 'TRANSACTION_CODE_ATB'],
    right_on=['APP_SOURCE_CODE', 'TRANSACTION_CODE_ATB'],
    how='left',
    indicator=True
).query("APP_SOURCE_CODE == 'MXPCP' and Categorie_de_transaction == 'code affectant directement le compte client (débit/crédit)'").copy(deep=True)
```
